# Week 7 — Neural Networks Fundamentals

## Neural Network

A neural network is a machine learning model that learns patterns from data and uses them to make predictions.

## Key Concepts

### 1. Layers
- **Input Layer:** Receives input data.
- **Hidden Layers:** Learn patterns from the data.
- **Output Layer:** Produces the final prediction.

### 2. Neurons
Neurons process input values using weights, biases, and activation functions.

### 3. Epochs
An epoch is one complete pass through the training dataset. Increasing epochs can improve learning, but too many may cause overfitting.

### 4. Activation Functions
Activation functions help neural networks learn complex patterns.

- **ReLU:** Commonly used in hidden layers.
- **Sigmoid:** Produces an output between 0 and 1.
- **Softmax:** Produces class probabilities for multi-class classification.

### 5. Loss Function
Measures the difference between the model's predictions and the actual results.

### 6. Optimizer
Updates the model's weights to reduce the loss. Adam is a commonly used optimizer.

### 7. Training and Evaluation
- **Training:** The model learns patterns from training data.
- **Evaluation:** Measures performance on unseen test data.
- **Overfitting:** Occurs when a model performs well on training data but poorly on new data.

## Conclusion

Neural networks learn by adjusting weights during training. Layers, neurons, epochs, activation functions, loss functions, and optimizers all influence model performance.
