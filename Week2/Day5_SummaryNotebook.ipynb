```python
import numpy as np
import pandas as pd
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Dimensions:", arr.ndim)
```

```python
print(arr[0])
print(arr[-1])
print(arr[1:4])
print(arr[::-1])
```

```python
print("Addition:", arr + 5)
print("Subtraction:", arr - 5)
print("Multiplication:", arr * 2)
print("Division:", arr / 2)
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])

reshaped = arr.reshape(2, 3)

print("Reshaped:")
print(reshaped)

print("Flattened:")
print(reshaped.flatten())
```

```python
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
```

```python
matrix1 = np.array([
    [1, 2],
    [3, 4]
])

matrix2 = np.array([
    [5, 6],
    [7, 8]
])

print("Addition:")
print(matrix1 + matrix2)

print("Multiplication:")
print(matrix1 * matrix2)

print("Matrix Multiplication:")
print(matrix1 @ matrix2)
```

```python
marks = pd.Series(
    [85, 72, 90, 65, 78],
    index=["Math", "Science", "English", "History", "Computer"]
)

print(marks)
print(marks["Math"])
print(marks[marks > 75])
```

```python
data = {
    "Name": ["Vijay", "Rahul", "John", "David"],
    "Age": [24, 25, 23, 26],
    "Marks": [85, 90, 78, 92]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.head())
print(df.tail())

df.info()

print(df.describe())
```

```python
print(df["Name"])

print(df[["Name", "Marks"]])

print(df[df["Marks"] > 80])
```

```python
df.loc[1, "Marks"] = np.nan

print(df.isnull().sum())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```

```python
df["Bonus"] = df["Marks"] + 5

print(df)
```

```python
data = {
    "Name": ["Vijay", "Rahul", "John", "David"],
    "Department": ["IT", "HR", "IT", "HR"],
    "Marks": [85, 90, 78, 92]
}

df = pd.DataFrame(data)

print(df.groupby("Department")["Marks"].mean())

print(df.groupby("Department")["Name"].count())
```

```python
data = {
    "Name": ["Vijay", "Rahul", "John", "David", "Mike"],
    "Math": [85, 90, np.nan, 92, 78],
    "Science": [80, 85, 75, np.nan, 82],
    "English": [90, 88, 82, 95, np.nan]
}

df = pd.DataFrame(data)

df["Math"] = df["Math"].fillna(df["Math"].mean())
df["Science"] = df["Science"].fillna(df["Science"].mean())
df["English"] = df["English"].fillna(df["English"].mean())

df["Total"] = df["Math"] + df["Science"] + df["English"]
df["Percentage"] = df["Total"] / 3

print(df)

print("\nAverage Percentage:", df["Percentage"].mean())
print("Highest Percentage:", df["Percentage"].max())
print("Lowest Percentage:", df["Percentage"].min())
```
