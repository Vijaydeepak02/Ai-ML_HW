## KMeans Clustering

Practiced grouping similar data points without predefined labels. Applied KMeans to student and customer datasets and inspected cluster assignments.

## Elbow Method

Tested different values of K and used the elbow plot to identify a suitable number of clusters.

## Feature Scaling

Compared clustering results before and after scaling to understand how differences in feature ranges affect distance-based algorithms.

## Principal Component Analysis (PCA)

Learned how PCA reduces the number of features while retaining important information. Transformed higher-dimensional data into two dimensions and inspected the explained variance ratio.

## Outlier Detection

Used the Interquartile Range (IQR) method and simple distance-based logic to identify unusual values. Compared results after removing or capping outliers.
